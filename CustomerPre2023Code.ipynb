```python
import pandas as pd
import os
from datetime import datetime

df = pd.read_csv('customer_database.csv')
```

```python
order1 = df[df['number_of_orders'] == 1]
order1 = order1[order1['most_used_email'].notna() & (order1['most_used_email'] != '')]
orderg100 = order1[order1['total_spent'] >= 100]
orderl100 = order1[order1['total_spent'] <= 100]
```

```python
orders_per_customer = orderl100

def extract_date(order_str):
    if pd.notna(order_str) and order_str:
        parts = order_str.split(' - ')
        if len(parts) > 0 and parts[0].startswith('Date:'):
            date_str = parts[0].replace('Date:', '').strip()
            try:
                return datetime.strptime(date_str, '%Y-%m-%d')
            except ValueError:
                return None
    return None

cutoff_date = datetime(2022, 12, 31)
def ordered_before_2023(row):
    order_col = f'order_1'
    if order_col in row and pd.notna(row[order_col]) and row[order_col]:
        order_date = extract_date(row[order_col])
        if order_date and order_date <= cutoff_date:
            return True
    return False

customers_before_2023 = orders_per_customer[orders_per_customer.apply(ordered_before_2023, axis=1)]

def get_earliest_order_date(row):
    dates = []
    order_col = f'order_1'
    if order_col in row and pd.notna(row[order_col]) and row[order_col]:
        order_date = extract_date(row[order_col])
        if order_date:
            dates.append(order_date)
    return min(dates)

customers_before_2023['earliest_order_date'] = customers_before_2023.apply(get_earliest_order_date, axis=1)

display_columns = ['first_name', 'last_name', 'most_used_email', 'number_of_orders', 'total_spent',
                   'earliest_order_date']

output_folder = 'CustomerPre2023'

if not os.path.exists(output_folder):
    os.makedirs(output_folder)
    
customers_before_2023[display_columns].to_csv(os.path.join(output_folder, 'customers_pre_2023_L100.csv'), index=False)
print(f"Filtered customers saved to {os.path.join(output_folder, 'customers_pre_2023_L100.csv')}")
```
